## Part III: Exploratory Data Analysis and Data Cleaning

The point of exploratory data analysis is to make sense of the data.

Exploratory data analysis encompasses activies like

1. Understanding the data structure
   *  look at the number of rows and columns to determine the scale of the data.
   *  look column labels, value types, and value ranges.
   *  identify categorical data and values taken by each kind of categorical data
   *  summarize the semantics of each column.

3. Summarizing data
   * computing summary statistics like means, stddevs, and counts.
     
4. Identifying missing values.
   * counting number of missing values and for which fields.
   * This can be used later in determining what kind of data cleaning may be needed.

4. Visualizing the data
   
6. Detecting outliers
   * create box plots of each type of numeric data. Using 1.5 IQR rule of thumb to
     identify outliers.
   * plot geographic data on a map or on a scatter plot to see if is sensical.

7. Assessing relationships
   * create correlation matrices
   * create contingency tables for categorical data
   * perform preliminary hypothesis tests, e.g., Chi-squared independence tests.

**Problem 3.A.9.** There are too many states to get an idea of how the customers are 
allocated geographically within Brazil.  Add a column to the DataFrame that denotes region
that is called `region`.
In Brazil states are grouped as follows:

* Southeast Region: Espírito Santo, Minas Gerais, Rio de Janeiro, and São Paulo
* South Region: Paraná, Santa Catarina, and Rio Grande do Sul
* North Region: Acre, Amapá, Amazonas, Pará, Rondônia, Roraima, and Tocantins
* Northeast Region: Alagoas, Bahia, Ceará, Maranhão, Paraíba, Pernambuco, Piauí, Rio Grande do Norte, and Sergipe
* Central-West Region: Goiás, Mato Grosso, Mato Grosso do Sul, and the Federal District.

Are there any customers that do not fall in one of the regions?

````python
from pyspark.sql.functions import when

state_to_region = {
    'ES': 'Southeast', 'MG': 'Southeast', 'RJ': 'Southeast', 'SP': 'Southeast',
    'PR': 'South', 'SC': 'South', 'RS': 'South',
    'AC': 'North', 'AP': 'North', 'AM': 'North', 'PA': 'North', 
    'RO': 'North', 'RR': 'North', 'TO': 'North',
    'AL': 'Northeast', 'BA': 'Northeast', 'CE': 'Northeast', 'MA': 'Northeast',
    'PB': 'Northeast', 'PE': 'Northeast', 'PI': 'Northeast', 'RN': 'Northeast', 'SE': 'Northeast',
    'GO': 'Central-West', 'MT': 'Central-West', 'MS': 'Central-West', 'DF': 'Central-West'
}

region_expr = when(col('customer_state') == 'ES', 'Southeast')
for state, region in list(state_to_region.items())[1:]:
    region_expr = region_expr.when(col('customer_state') == state, region)
region_expr = region_expr.otherwise('Unknown')

customers_df = customers_df.withColumn('region', region_expr)
unknown_count = customers_df.filter(col('region') == 'Unknown').count()
print(f"Customers not in any region: {unknown_count}")
display(customers_df.limit(10))
````

#### Inspecting Driver Memory

There are several ways to inspect the memory in the driver node including.

```
  psutil.virtual_memory()
```

and

```
  pages = os.sysconf("SC_PHYS_PAGES")
  page_size = os.sysconf("SC_PAGE_SIZE")
  total_mem = pages * page_size
```

The function `psutil.virtual_memory()` is poorly named because it returns an object with many fields providing information about the memory of the driver node.  Which fields are present depends on the operating system.   See https://psutil.readthedocs.io/latest/api.html#memory

ASIDE: I found there was more than enough memory in the driver, which makes it easier to 
visualize the raw data without having to employ special considerations for outliers.


````python
import psutil

sample_fraction = 0.01
sample_df = geo_df.sample(fraction=sample_fraction).toPandas()
sample_memory = sample_df.memory_usage(deep=True).sum()
estimated_full_memory = sample_memory / sample_fraction
vm = psutil.virtual_memory()
print(f"sample memory usage: {sample_memory / (1024*1024):.2f} MB")
print(f"full dataset memory: {estimated_full_memory / (1024*1024):.2f} MB")
print(f"available memory: {vm.available / (1024*1024):.2f} MB")
print(f"total memory: {vm.total / (1024*1024):.2f} MB")
````

It is safe to call toPandas() on the full dataset since the estimated memory requirement a very small portion of the available memory.

The main assumption when extrapolating from the sample is that the sample is representative of the full dataset for string lengths and data types. This estimate could be inaccurate if the sample contains short or long strings compared to the rest of the data.

````python
orders_df = spark.createDataFrame(
    pd.read_csv(data_path + 'olist_orders_dataset.csv')
)

order_items_df = spark.createDataFrame(
    pd.read_csv(data_path + 'olist_order_items_dataset.csv')
)

order_reviews_df = spark.createDataFrame(
    pd.read_csv(data_path + 'olist_order_reviews_dataset.csv')
)
````

````python
from pyspark.sql.functions import sum as spark_sum, to_timestamp, datediff
from scipy import stats

orders_customers = orders_df.join(customers_df, 'customer_id')
orders_items_customers = orders_customers.join(order_items_df, 'order_id')

sp_orders = orders_items_customers.filter(col('customer_state') == 'SP')\
    .groupBy('order_id').agg(spark_sum('price').alias('total_price'))
rj_orders = orders_items_customers.filter(col('customer_state') == 'RJ')\
    .groupBy('order_id').agg(spark_sum('price').alias('total_price'))

sp_pd = sp_orders.toPandas()['total_price']
rj_pd = rj_orders.toPandas()['total_price']

t_stat, p_value = stats.ttest_ind(sp_pd, rj_pd)
print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")
print(f"SP mean: {sp_pd.mean():.2f}")
print(f"RJ mean: {rj_pd.mean():.2f}")
````

The p-value = 0.0000 < 0.01, so we reject the null hypothesis. Rio de Janeiro (mean = $142.93) has significantly higher average purchase amounts than São Paulo (mean = $125.75).

````python
import numpy as np

observed_diff = rj_pd.mean() - sp_pd.mean()
combined = np.concatenate([sp_pd, rj_pd])
n_sp = len(sp_pd)

n_permutations = 1000
perm_diffs = []
for _ in range(n_permutations):
    np.random.shuffle(combined)
    perm_diffs.append(combined[n_sp:].mean() - combined[:n_sp].mean())

p_value_perm = np.mean(np.abs(perm_diffs) >= np.abs(observed_diff))

plt.figure(figsize=(10, 5))
plt.hist(perm_diffs, bins=30)
plt.axvline(observed_diff, color='red', linewidth=2, label=f'Observed diff = {observed_diff:.2f}')
plt.xlabel('Difference in Means')
plt.ylabel('Frequency')
plt.title('Permutation Test: SP vs RJ Purchase Amounts')
plt.legend()
plt.show()

print(f"Observed difference: {observed_diff:.4f}")
print(f"Permutation p-value: {p_value_perm:.4f}")
````

The permutation p-value is 0.0000, matching the t-test. The observed difference of 17.18 is far outside the permutation distribution, so the result is the same. RJ spends significantly more than SP.

````python
from pyspark.sql.functions import to_timestamp, datediff

orders_reviews = orders_df.join(order_reviews_df, 'order_id')

orders_reviews = orders_reviews.withColumn(
    'delivery_time',
    datediff(
        to_timestamp(col('order_delivered_customer_date')),
        to_timestamp(col('order_purchase_timestamp'))
    )
)

orders_reviews_pd = orders_reviews.select('delivery_time', 'review_score')\
    .dropna().toPandas()

r, p_value = stats.pearsonr(orders_reviews_pd['delivery_time'], orders_reviews_pd['review_score'])
print(f"Pearson r: {r:.4f}")
print(f"P-value: {p_value:.4f}")
````

r = -0.3337, p-value = 0.0000. We reject the null hypothesis, there is a statistically significant negative correlation between delivery time and review score. Longer deliveries are associated with lower review scores.

````python
observed_r = r
delivery = orders_reviews_pd['delivery_time'].values
scores = orders_reviews_pd['review_score'].values

n_permutations = 1000
perm_rs = []
for _ in range(n_permutations):
    shuffled = np.random.permutation(scores)
    perm_r, _ = stats.pearsonr(delivery, shuffled)
    perm_rs.append(perm_r)

p_value_perm = np.mean(np.abs(perm_rs) >= np.abs(observed_r))

plt.figure(figsize=(10, 5))
plt.hist(perm_rs, bins=30)
plt.axvline(observed_r, color='red', linewidth=2, label=f'Observed r = {observed_r:.4f}')
plt.xlabel('Pearson r')
plt.ylabel('Frequency')
plt.title('Permutation Test: Delivery Time vs Review Score')
plt.legend()
plt.show()

print(f"Observed r: {observed_r:.4f}")
print(f"Permutation p-value: {p_value_perm:.4f}")
````
